```python
from diffusers import StableDiffusionPipeline
import torch
from IPython.display import display

model_id = "runwayml/stable-diffusion-v1-5"

# Load model and prepare pipeline
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda")
pipe.enable_attention_slicing() # helps on smaller GPUs

# -------------------------------
# Define your text prompt
# (Pick ONE from the ideas below or write your own)
# -------------------------------
prompt = "cyberpunk alleyway in the rain, digital art style, reflections and neon signs"
# Examples you can swap in:
# prompt = "a serene forest with mist and sunlight rays, ultra-realistic, detailed foliage"
# prompt = "a cyberpunk alleyway in the rain, digital art style, reflections and neon signs"
# prompt = "a fantasy castle on a mountain, oil painting, golden hour lighting, dramatic
# clouds"
# prompt = "a portrait of a woman in Van Gogh’s style, textured brush strokes, vivid tones"

# Generate the image
result = pipe(
prompt,
num_inference_steps=30, # number of denoising steps
guidance_scale=7.5, # how closely it follows your text
height=512, width=512 # image size; keep low for limited VRAM
)

# Display and save
image = result.images[0]
display(image) # shows image inside Colab
image.save("generated.png") # save locally
print("Saved to generated.png")
```

```python
# Helper: generate, display, and save with a fixed seed for reproducibility
import torch
from IPython.display import display
import os

os.makedirs("ex01_outputs", exist_ok=True)

def run_and_save(label, prompt, filename,
    steps=30, scale=7.5, h=512, w=512, seed=1234,
    negative_prompt=None):
    generator = torch.Generator(device="cuda").manual_seed(seed)
    out = pipe(
        prompt,
        num_inference_steps=steps,
        guidance_scale=scale,
        height=h, width=w,
        negative_prompt=negative_prompt,
        generator=generator
    )
    img = out.images[0]
    print(f"[{label}] Prompt:\n{prompt}\n")
    display(img)
    path = f"ex01_outputs/{filename}"
    img.save(path)
    print(f"Saved -> {path}\n")
    return img
```

```python
# -------------------------------
# Weak Prompt
# -------------------------------
prompt_weak = "a bird"
img_weak = run_and_save("WEAK", prompt_weak, "01_weak_bird.png")

# -------------------------------
# Strong Prompt
# -------------------------------
prompt_strong = (
"a colorful macaw flying over a rainforest in morning light, "
"realistic, 4K, detailed feathers, shallow depth of field, vibrant colors"
)
img_strong = run_and_save("STRONG", prompt_strong, "02_strong_macaw.png")

neg = "blurry, low quality, watermark, text, deformed, extra limbs"
img_strong_neg = run_and_save(
    "STRONG+NEG",
    prompt_strong,
    "03_strong_macaw_neg.png",
    negative_prompt=neg
)

# (Optional) Sensitivity test change ONE word
prompt_variant = (
    "a colorful macaw flying over a rainforest in golden hour light, "
    "realistic, 4K, detailed feathers, vibrant colors"
)
img_variant = run_and_save("VARIANT", prompt_variant, "04_variant_macaw.png")
```

```python
# -------------------------------
# Base Prompt (equal emphasis)
# -------------------------------
prompt_base = "a futuristic street at night with neon lights, cinematic lighting"
img_base = run_and_save("BASE", prompt_base, "01_base_street.png")

# -------------------------------
# Weighted Emphasis: NEON lights
# -------------------------------
prompt_weighted1 = (
"a futuristic street at night with bright NEON lights, "
"glowing NEON signs, NEON reflections, cinematic lighting, high contrast"
)
img_weighted1 = run_and_save("NEON EMPHASIS", prompt_weighted1, "02_neon_emphasis.png")

# -------------------------------
# Weighted Emphasis: STREET detail
# -------------------------------
prompt_weighted2 = (
"a detailed futuristic STREET scene, wide STREET perspective, "
"STREET reflections, dim NEON lights in background, atmospheric fog"
)
img_weighted2 = run_and_save("STREET EMPHASIS", prompt_weighted2, "03_street_emphasis.png")

# -------------------------------
# Compare stronger guidance
# -------------------------------
prompt_strong_guidance = (
"a futuristic street at night glowing with NEON lights and reflections, "
"cinematic lighting, hyper-realistic"
)
img_strong_guidance = run_and_save(
"HIGH GUIDANCE", prompt_strong_guidance, "04_high_guidance.png",
scale=9.0 # higher adherence to text
)
```
